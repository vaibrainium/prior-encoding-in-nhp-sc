# GLM-HMM state schematics
Schematic psychometric curves for the left-bias, right-bias and unbiased states.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

OUT = Path.cwd().parent / "dissemination" / "glm_hmm_states"  # notebooks/ is the cwd
OUT.mkdir(parents=True, exist_ok=True)

# name: (color, slope, midpoint, lower asymptote, upper asymptote)
STATES = {
    "left_bias": ("#E8749B", 1.6, 0.8, 0.0, 0.95),
    "right_bias": ("#377eb8", 1.6, -0.8, 0.05, 1.0),
    "unbiased": ("#ff7f00", 4.0, 0.0, 0.02, 0.98),
}
x = np.linspace(-3, 3, 300)
```

```python
def curve(slope, mid, lo, hi):
    return lo + (hi - lo) / (1 + np.exp(-slope * (x - mid)))
```

```python
def draw(ax, name, lw=2.5):
    color, *p = STATES[name]
    ax.plot(x, curve(*p), color=color, lw=lw, solid_capstyle="round")
```

```python
def style(ax):
    ax.figure.set_facecolor("#E6E6E6")
    ax.set_facecolor("#E6E6E6")
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_ylim(-0.05, 1.05)
    #x and y axis thicker
    ax.spines["left"].set_linewidth(2)
    ax.spines["bottom"].set_linewidth(2)
    for side in ("top", "right"):
        ax.spines[side].set_visible(False)
    ax.spines["left"].set_color("#555")
    ax.spines["bottom"].set_color("#555")
```

```python
# individual panels
for name, (color, *_) in STATES.items():
    fig, ax = plt.subplots(figsize=(2, 2))
    style(ax)
    draw(ax, name,lw=5)
    plt.show()
```
